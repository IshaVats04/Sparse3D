## Step 5: Verify & Render Dense Point Cloud (`fused.ply`)

```python
import open3d as o3d
from pathlib import Path

fused_ply = Path('outputs/dtu_scan1_6views_lighting0/dense/fused.ply')
if fused_ply.is_file():
    pcd = o3d.io.read_point_cloud(str(fused_ply))
    print(f'✅ Dense Point Cloud Successfully Generated!')
    print(f'Total Number of 3D Points: {len(pcd.points):,}')
    print(f'Bounding Box Min: {pcd.get_min_bound()}')
    print(f'Bounding Box Max: {pcd.get_max_bound()}')
else:
    print('❌ fused.ply not found. Check pipeline output logs.')
```
